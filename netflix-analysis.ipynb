## Data Cleaning

```python
df['country'] = df['country'].fillna('Unknown')
```

```python
df['date_added'].isna().mean()*100
```

```python
df = df.dropna(subset = ['date_added'])
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
len(df[ df['type'] == 'Movie'])
# or df['type'].value_counts()
```

```python
len(df[ df['type'] == 'TV Show'])
#  or df['type'].value_counts()
```

## Time-based Analysis

## Rating Analysis

```python
df['rating'].value_counts()
```

```python
df['rating'].value_counts().plot(kind='bar', figsize=(10, 6))
plt.title('Distribution of Rating')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
df['rating'].value_counts().sort_values(ascending=False).head(5)
```

## Duration Analysis

```python
movies = df[df['type'] == 'Movie'].copy()
movies['duration_int'] = movies['duration'].str.extract('(\d+)').astype(float)
movies['duration_int'].mean()
```

```python
movies.loc[movies['duration_int'].idxmax()]
# longest movie
```

```python
movies.loc[movies['duration_int'].idxmin()]
# shortest movie
```
